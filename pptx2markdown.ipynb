### PPTX2MD Batch Automation Code

To achieve your goals, pptx2md is the better engine for your project.  

Microsoft's markitdown focuses primarily on extracting text and text metadata for AI pipelines; it does not natively extract and save raw presentation image files into a local folder.


The pptx2md library fulfills all 5 of your requirements out of the box. The complete automation script below can be run inside a single Google Colab cell.


```python
# Step 1: Install the conversion tool
!pip install pptx2md

import os
import glob
from google.colab import drive
```

check result

```python
# Step 2: Connect to your Google Drive
# You will be prompted to grant permissions
drive.mount('/content/drive')

# -------------------------------------------------------------
# CONFIGURATION: Change 'My_PPTX_Folder' to your actual folder name
# -------------------------------------------------------------
# FOLDER_NAME = "My_PPTX_Folder"
FOLDER_NAME = "SDGAI/C3659C/"
SOURCE_DIR = f"/content/drive/MyDrive/{FOLDER_NAME}"
# -------------------------------------------------------------
```

check result

```python
# Find all PowerPoint files in the directory
pptx_files = glob.glob(os.path.join(SOURCE_DIR, "*.pptx"))

if not pptx_files:
    print(f"No .pptx files found in: {SOURCE_DIR}")
    print("Please double-check your folder name.")
else:
    print(f"Found {len(pptx_files)} presentation(s). Starting batch conversion...\n")

    for pptx_path in pptx_files:
        # Get the file name without extension (e.g., "presentation")
        base_name = os.path.splitext(os.path.basename(pptx_path))[0]

        # Define paths according to your requirements
        output_md_path = os.path.join(SOURCE_DIR, f"{base_name}.md")
        image_subfolder = os.path.join(SOURCE_DIR, f"{base_name}_images")

        print(f"Converting: {base_name}.pptx")
        print(f"└─ Images saving to sub-folder: {base_name}_images/")

        # Step 3, 4, & 5: Run conversion with target names and sub-folders
        # -o specifies the output filename
        # --image-dir specifies the sub-folder for images
        command = f'pptx2md "{pptx_path}" -o "{output_md_path}" --image-dir "{image_subfolder}"'
        os.system(command)

    print("\n All conversions completed successfully! Check your Google Drive folder.")
```
